## 1. Вектор

Вектор — одномерная структура данных. В Python её можно представить как:
- обычный `list`,
- `numpy.ndarray` (для вычислений),
- `pandas.Series` (с метками).

```python
# 1.1 Создаём вектор S1 со значениями от 1 до 10 всеми возможными способами

# Способ 1: обычный список Python
S1_list = list(range(1, 11))
print("Способ 1 (list):", S1_list)

# Способ 2: NumPy-массив
S1 = np.arange(1, 11)
print("Способ 2 (np.arange):", S1)
print("Тип:", type(S1))

# Способ 3: Pandas Series
S1_series = pd.Series(range(1, 11))
print("Способ 3 (pd.Series):")
print(S1_series)
```

```python
# 1.2 Вектор S2 от 1 до 10 с шагом 0.5
S2 = np.arange(1, 10.5, 0.5)     # 19 значений: 1.0, 1.5, ..., 10.0
# Альтернатива: S2 = np.linspace(1, 10, 19)
print("S2:", S2)
print("Длина S2:", len(S2))
```

```python
# 1.3 Объединяем S1 и S2 в вектор S (в R это c(S1, S2))
S = np.concatenate([S1, S2])
print("Объединённый вектор S:")
print(S)
print("Длина S:", len(S))
```

```python
# 1.4 Выводим 3-е, 4-е и 5-е значения вектора S
# Python индексирует с 0 -> индексы 2, 3, 4
print("3-е, 4-е, 5-е значения S:", S[[2, 3, 4]])
```

```python
# 1.5 Первое и четвёртое значения вектора S
print("1-е и 4-е значения S:", S[[0, 3]])
```

## 2. Таблица (DataFrame)

Основная табличная структура в Python — `pandas.DataFrame`.

```python
# 2.1 Создаём текстовые векторы City и Sex и вектор Number
MaleStruct = ["Female", "Male"]
Cities = ["Владивосток", "Москва", "СПБ", "Челябинск", "Ульяновск", "Казань"]
Number = [12000, 8000, 15000, 5000, 20000, 22000]

print("Cities:", Cities)
print("MaleStruct:", MaleStruct)
print("Number:", Number)

# Собираем столбец Sex, повторяя MaleStruct до нужной длины
Sex = (MaleStruct * 3)[:len(Number)]
print("Sex:", Sex)
```

```python
# 2.3 Объединяем три вектора в таблицу CITY с заголовками
CITY = pd.DataFrame({
    "City": Cities,
    "Sex": Sex,
    "Number": Number
})

# 2.4 Визуализация содержимого таблицы
CITY
```

```python
# 2.5 Имена переменных (столбцов) таблицы CITY
print(CITY.columns.tolist())
```

```python
# 2.6 Извлекаем элементы 1–3 из столбца Number
print(CITY["Number"].iloc[0:3])
```

```python
# 2.7 Значения численности, превышающие 10000
CITY[CITY["Number"] > 10000]
```

```python
# 2.8 Значения численности мужского населения
CITY[CITY["Sex"] == "Male"]
```

```python
# 2.9 Первые 3 и последние 2 строки таблицы
print("Первые 3 строки:")
display(CITY.head(3))

print("Последние 2 строки:")
display(CITY.tail(2))
```

## 3. Матрица

В Python матрица — это `numpy.ndarray` с `ndim == 2`.

```python
# 3.1 Матрица размерности 2×8
m1 = np.arange(1, 17).reshape(2, 8)
print("m1 (2×8):")
print(m1)
print("Форма:", m1.shape)
```

```python
# 3.2 Числовая матрица размерности 4×6
m3 = np.arange(1, 25).reshape(4, 6)
print("m3 (4×6):")
print(m3)
print("Форма:", m3.shape)
```

```python
# 3.3 Транспонируем матрицу m3
m3_T = m3.T
print("m3_T (6×4):")
print(m3_T)
print("Форма:", m3_T.shape)
```

```python
# 3.4 Квадратная матрица и её определитель
square = m3 @ m3_T     # 4×4
print("Квадратная матрица:")
print(square)

det = np.linalg.det(square)
print("Определитель:", det)
```

## 4. Массив

В R `array()` создаёт массив любой размерности. В Python — тот же `numpy.ndarray`.

```python
# 4.1 Массив M из 24 элементов
M = np.array(range(1, 25)).reshape(3, 2, 4)   # 3×2×4 = 24
print("Форма M:", M.shape)
print(M)
```

```python
# 4.2 Проверяем, является ли M массивом
print("isinstance(M, np.ndarray):", isinstance(M, np.ndarray))
print("ndim:", M.ndim)
print("shape:", M.shape)
print("size:", M.size)
```

## 5. Список

В R `list` может хранить элементы разных типов и длин.  
В Python для «именованного списка» удобно использовать `dict`.

```python
# 5.1 Создаём список L1: числа 1..6 и (TRUE, FALSE) три раза
integers = list(range(1, 7))
booleans = [True, False] * 3

L1 = [integers, booleans]
print("L1 (без имён):")
print(L1)
```

```python
# 5.2 Присваиваем имена элементам списка
L1 = {
    "integers": integers,
    "booleans": booleans
}
print("L1 (с именами):")
print(L1)
```

## 6. Фактор

Аналог R-фактора в Python — `pandas.Categorical`.

```python
# 6.1 Вектор из 6 объектов трёх классов: Yes, No, Parhaps -> фактор
answers = ["Yes", "No", "Parhaps", "Yes", "No", "Parhaps"]

f1 = pd.Categorical(answers, categories=["Yes", "No", "Parhaps"])
print("f1:")
print(f1)
print("Уровни:", f1.categories.tolist())
print("Коды:", f1.codes)
```

## 7. Фрейм данных

Создаём `DataFrame` из разнотипных векторов, включая фактор.

```python
# 7.1 Фрейм F1
b = ["a", "b", "c", "a"]
d = [(i % 2 == 0) for i in range(1, 5)]     # [False, True, False, True]
e = pd.Categorical(["soft", "hard", "soft", "medium"],
                   categories=["soft", "hard", "medium"])

F1 = pd.DataFrame({
    "b": b,
    "d": d,
    "e": e
})

F1
```
